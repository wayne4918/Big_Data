범죄 데이터 정리 및 정규화 코드

```python
from sklearn.preprocessing import MinMaxScaler
import pandas as pd

# 범죄 데이터 불러오기
crime_data_path = "/content/drive/My Drive/bigdata/빅데이터 기말과제 데이터/5대+범죄+발생현황_20241220144949.xlsx"
crime_data = pd.read_excel(crime_data_path)

# 필요한 열 선택
crime_columns = ['자치구별(2)', '2022', '2022.1', '2022.2', '2022.3', '2022.4', '2022.5']
crime_data_cleaned = crime_data[crime_columns].copy()

# 숫자로 변환 (비숫자 값 제거 및 NaN 처리)
for column in crime_columns[1:]:  # 자치구 이름은 변환하지 않음
    crime_data_cleaned[column] = pd.to_numeric(crime_data_cleaned[column], errors='coerce')

# '소계'와 같은 집계 행 제거
crime_data_cleaned = crime_data_cleaned[~crime_data_cleaned['자치구별(2)'].str.contains("소계")]

# 결측값 제거
crime_data_cleaned = crime_data_cleaned.dropna()

# 정규화 수행
scaler = MinMaxScaler()
crime_data_cleaned.iloc[:, 1:] = scaler.fit_transform(crime_data_cleaned.iloc[:, 1:])

# 결과 확인
print("Normalized Crime Data:")
print(crime_data_cleaned.head())

# 정리된 데이터 저장
crime_data_cleaned.to_csv('crime_normalized.csv', index=False, encoding='utf-8-sig')
print("Normalized crime data has been saved to 'crime_normalized.csv'.")
```

```python
import pandas as pd
import numpy as np

# 1. 자치구별 집값 데이터 로드
house_price_path = "/content/drive/My Drive/bigdata/빅데이터 기말과제 데이터/자치구별 집값.xlsx"
house_price_data = pd.read_excel(house_price_path)

# 2. 데이터 확인 및 필요한 컬럼 선택
print(house_price_data.head())  # 데이터 확인
# 필요한 컬럼만 선택 ('자치구'와 '가격평균' 관련 컬럼으로 가정)
house_price_cleaned = house_price_data[['자치구', '가격평균']].copy()

# 3. 데이터 클리닝 (결측치 제거)
house_price_cleaned.dropna(inplace=True)

# 4. 로그 변환 적용
house_price_cleaned['가격평균_로그'] = np.log1p(house_price_cleaned['가격평균'])  # log(1 + x) 변환

# 5. 결과 확인
print("House Price Data (Log Transformed):")
print(house_price_cleaned.head())

# 6. 정리된 데이터 저장
house_price_cleaned.to_csv('house_price_with_log.csv', index=False, encoding='utf-8-sig')
print("House price data with log transformation has been saved to 'house_price_with_log.csv'.")

# 7. 통합 데이터와 병합
# (통합 데이터는 기존의 merged_data와 연결)
merged_data = pd.read_csv('merged_data.csv')  # 기존 통합 데이터 로드
final_data_with_house_price = pd.merge(
    merged_data,
    house_price_cleaned[['자치구', '가격평균_로그']],
    on='자치구',
    how='left'
)

# 8. 최종 데이터 저장
final_data_with_house_price.to_csv('final_data_with_log_price.csv', index=False, encoding='utf-8-sig')
print("Final data with log-transformed house price has been saved to 'final_data_with_log_price.csv'.")
```

```python
import pandas as pd

# 1. 정규화된 데이터 불러오기
cctv_data = pd.read_csv('cctv_normalized.csv')
medical_data = pd.read_csv('medical_cleaned.csv')
crime_data = pd.read_csv('crime_normalized.csv')
management_data = pd.read_csv('management_normalized.csv')
road_data = pd.read_csv('road_normalized.csv')
culture_data = pd.read_csv('culture_normalized.csv')
police_fire_data = pd.read_csv('police_fire_normalized.csv')

# 2. 자치구 컬럼 통일
# CCTV 데이터
if '구분' in cctv_data.columns:
    cctv_data.rename(columns={'구분': '자치구'}, inplace=True)

# 의료기관 데이터
if '자치구별(2)' in medical_data.columns:
    medical_data.rename(columns={'자치구별(2)': '자치구'}, inplace=True)

# 범죄 데이터
if '자치구별(2)' in crime_data.columns:
    crime_data.rename(columns={'자치구별(2)': '자치구'}, inplace=True)

# 관리비 데이터
if '지역' in management_data.columns:
    management_data.rename(columns={'지역': '자치구'}, inplace=True)

# 도로 시설물 데이터
if '자치구' not in road_data.columns:
    road_data.rename(columns={'자치구별(2)': '자치구'}, inplace=True)

# 문화공간 데이터
if '자치구별(2)' in culture_data.columns:
    culture_data.rename(columns={'자치구별(2)': '자치구'}, inplace=True)

# 경찰·소방 데이터
if '자치구별(2)' in police_fire_data.columns:
    police_fire_data.rename(columns={'자치구별(2)': '자치구'}, inplace=True)

# 3. '자치구' 컬럼 정리 (공백, 특수문자 제거)
datasets = [cctv_data, medical_data, crime_data, management_data, road_data, culture_data, police_fire_data]
for df in datasets:
    df['자치구'] = df['자치구'].str.strip()  # 공백 제거

# 4. 데이터 병합
merged_data = datasets[0]
for df in datasets[1:]:
    merged_data = pd.merge(merged_data, df, on='자치구', how='inner')

# 5. 결과 확인
print("Merged DataFrame:")
print(merged_data.head())

# 병합된 데이터를 CSV 파일로 저장
merged_data.to_csv('merged_data.csv', index=False, encoding='utf-8-sig')

print("병합된 데이터가 'merged_data.csv'로 저장되었습니다.")
```

시사점
독거노인 문제 대응 시, 시기별 정책 변화와 사회적 요인을 함께 고려해야 함.
예를 들어, 전염병 사태 시 독거노인을 위한 긴급 복지 지원, 경기침체 시 주거·생활 지원 정책 강화 등.
2. 독거노인 연령대별 상호 높은 상관관계
주요 관찰
65~79세, 80세 이상 독거노인 집단 간에 높은 양의 상관관계를 나타냄.
한 연령대에서 독거노인이 증가하면, 다른 연령대에서도 유사한 증가 추이를 보임.
시사점
독거노인 문제가 특정 연령대만의 문제가 아니라, 고령층 전반의 구조적 문제일 수 있음.
정책적으로 한 연령대를 초점화하기보다, 고령층 전체를 아울러 지원하는 통합적 접근이 필요.
3. 시설·인프라 변수와의 낮거나 음(陰)의 상관관계
주요 관찰
CCTV, 경찰서, 소방서, 복지관, 체육시설 등 다양한 인프라가 독거노인 수와 대부분 낮거나 음의 상관을 보임.
이는 단순 인프라 확대가 독거노인 수나 삶의 질 개선과 직결되지 않는다는 것을 의미.
시사점
양적인 인프라 구축만으로는 한계가 있음. 실질적인 접근성과 서비스 질을 높이는 것이 필요함.
예: 복지관이나 체육시설의 물리적 배치뿐 아니라, 독거노인이 실제로 방문·활용하기 편하도록 제도·교통편·프로그램 등을 마련해야 함.
4. 생활환경·경제적 지표와의 음의 상관관계
관리비, 장기수선충당금 등

관리비가 높을수록, 장기수선충당금이 높을수록 독거노인 수와 음의 상관이 관찰됨.
높은 주거비용은 독거노인에게 부담이 되어, 해당 지역에서 거주하기 어렵다는 가능성을 시사.
가격평균

지역별 평균 가격이 높을수록 독거노인의 분포가 낮아지는 경향(음의 상관관계)이 뚜렷함.
이는 주택가격, 생활비 등이 높은 지역에서 독거노인이 거주하기 어렵다는 구조적 문제를 보여줌.
시사점
독거노인의 경제적 취약성이 부각됨에 따라, 저소득층 지원, 임대료 보조, 관리비 완화 등이 중요 과제로 떠오름.
단순 “가격평균이 높다/낮다”가 아니라, 그 가격 수준을 상대적으로(정규화) 보아 지역 간 불균형이 심한 곳에 집중 지원할 필요가 있음.
5. 종합 결론 요약
시간적·사회적 요인에 대한 대응

시기별(연도별, 특정 이벤트 시점)로 독거노인 수가 큰 폭으로 변동하므로, 주기적 모니터링과 유연한 정책 설계가 필수.
연령대 포괄적 접근

독거노인 문제는 고령층 전체의 문제로 확대 해석 가능.
모든 고령층을 아울러 지원·보호하는 통합적 복지·의료·돌봄 시스템 필요.
인프라 ‘양적 확대’의 한계

단순히 시설만 늘리는 것으로는 독거노인 삶의 질 개선에 한계가 있음.
질적 서비스, 접근성 강화, 맞춤형 프로그램 등을 통해 실질적 개선이 이루어져야 함.
비용 장벽 해소

높은 관리비, 주거 비용(가격평균, 가격평균_정규화 등)은 독거노인이 거주하기 어려운 주요 요인.
**경제적 지원정책(주거비 지원, 세제 혜택, 공동생활 지원 등)**을 강화하여, 독거노인이 안정적으로 거주할 수 있는 환경을 조성해야 함.
마무리
독거노인 문제는 단순히 한두 가지 원인(인프라 부족, 주거비 부담 등)으로 설명하기 어렵고, 시기적·경제적·사회적 요인이 결합된 복합 문제임을 상관계수 분석 결과가 보여줍니다.
특히, **가격평균(및 정규화된 가격지표)**이 높은 지역일수록 독거노인 수가 줄어드는 경향은, 경제적 요인이 가장 직접적이라는 점을 다시금 강조합니다.
결국, 포괄적 복지정책(인프라 접근성, 경제 지원, 시기별 대응, 연령대별 맞춤 서비스)을 동시다발적으로 추진해야만, 독거노인 문제에 효과적으로 대응할 수 있을 것입니다.

```python
import pandas as pd
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt

# 1. 데이터 로드
merged_data = pd.read_csv('final_data_with_price_and_log.csv')  # 통합 데이터
lonely_senior_data = pd.read_csv('lonely_senior_cleaned.csv')  # 독거노인 데이터

# 2. 데이터 병합 (자치구 기준)
final_data = pd.merge(merged_data, lonely_senior_data, on='자치구', how='inner')

# 3. 숫자형 데이터만 추출
numeric_data = final_data.select_dtypes(include=['float64', 'int64'])

# 4. 데이터 정규화
scaler = StandardScaler()
scaled_data = scaler.fit_transform(numeric_data)

# 5. PCA 적용
pca = PCA(n_components=4)  # 주요 4개 요인 추출
principal_components = pca.fit_transform(scaled_data)

# 6. PCA 결과 데이터프레임 생성
pca_columns = ['안전성', '편의성', '비용 효율성', '집값']
pca_data = pd.DataFrame(data=principal_components, columns=pca_columns)

# 7. 가중치를 적용한 종합 점수 계산
weights = {'안전성': 0.25, '편의성': 0.25, '비용 효율성': 0.25, '집값': -0.25}  # 집값에 음의 가중치 적용
pca_data['종합 점수'] = (
    pca_data['안전성'] * weights['안전성'] +
    pca_data['편의성'] * weights['편의성'] +
    pca_data['비용 효율성'] * weights['비용 효율성'] +
    pca_data['집값'] * weights['집값']
)

# 8. 자치구 컬럼 추가
pca_data['자치구'] = final_data['자치구']

# 9. 결과 저장
pca_data.to_csv('final_data_with_pca_and_house_price.csv', index=False)
print("PCA 결과와 집값 점수가 포함된 데이터가 'final_data_with_pca_and_house_price.csv'에 저장되었습니다.")

# 10. 결과 시각화
plt.figure(figsize=(10, 6))
plt.bar(pca_data['자치구'], pca_data['종합 점수'], color='skyblue')
plt.xlabel('자치구', fontsize=12)
plt.ylabel('종합 점수', fontsize=12)
plt.title('자치구별 종합 점수 ', fontsize=16)
plt.xticks(rotation=45, fontsize=10)
plt.tight_layout()
plt.show()


# 12. PCA 설명력 출력
print("PCA 설명력 (각 주성분):", explained_variance_ratio[:4])
print("총 설명력 (4개 주성분):", cumulative_variance[3])
```

1. 상관관계 표(히트맵)에서의 주요 포인트
안전성 ↔ 종합점수 (상관계수 0.71)

안전성 지표가 높을수록 종합 점수가 높아지는 경향이 가장 뚜렷합니다.
즉, ‘안전’ 요소가 최종 평가(종합 점수)에 큰 비중을 차지하고 있음을 시사합니다.


편의성이 높아질수록 오히려 집값은 낮은 경향을 보이는 음(陰)의 상관관계입니다.
일반적으로 “편의성이 높은 곳 = 집값이 비싸다”는 통념과 다른 결과일 수 있으며,
실제로는 ‘상권·시설이 잘 갖춰져 있지만 상대적으로 집값 상승 요인이 적은 지역’이 존재하거나,
데이터 처리・평가 기준상 편의성이란 항목이 집값과 반비례하도록 산출되었을 가능성도 고려해야 합니다.


비용 효율성이 높은 지역일수록 집값도 높게 나타나는 양의 상관관계입니다.
“비용 효율성”을 단순히 ‘집값이 싼가?’로만 보지 않고, **해당 집값 수준 대비 얻을 수 있는 장점(교통, 인프라 등)**을 고려한 지표라면,
‘집값은 높지만, 그만큼 가치(효용)도 높다’라는 의미로 해석될 수 있습니다.


집값과 종합 점수 사이의 상관관계는 비교적 낮습니다.
집값이 비싸다고 해서 종합 점수가 반드시 높아지지 않는다는 점을 보여줍니다(안전성·편의성 등 다른 요소의 영향이 더 큼).


2. 자치구별 안전성·편의성·비용 효율성 점수 분포
그래프를 통해 확인할 수 있는 특징

어떤 자치구는 안전성이 높지만 편의성·비용 효율성이 상대적으로 낮을 수 있고,
반대로 편의성이 높게 나타나지만 안전성은 낮은 식의 특화된 장단점이 존재합니다.
집값과 점수들의 상관이 명확히 일치하지 않으므로, 무조건 집값이 비싸다고 모든 점수가 우수한 것은 아님을 알 수 있습니다.
정책·의사결정 관점 시사점

지역별 특성(안전 위주 강화, 편의시설 확충, 비용 대비 효용)을 균형감 있게 고려해야 함.
특정 자치구가 ‘안전’은 좋은데 ‘편의성’이 부족하다면, 편의시설 개선을 우선 과제로 삼을 수 있고,
반대로 편의성과 비용 효율성이 높은 지역이라면, 주거·치안 등에 집중 투자해 균형을 맞출 수도 있습니다.
3. 종합 결론
안전성의 절대적 영향

전체 평가(종합 점수)에서 가장 높은 상관을 보이는 항목은 ‘안전성’입니다.
안전 문제(치안, 재난 대응 등)를 개선하는 것이 종합적 삶의 질 및 지역 평가를 높이는 데 핵심 요인임을 시사합니다.
편의성 vs. 집값의 음(陰)의 상관관계

편의성과 집값이 부(-)의 상관을 보인다는 점이 흥미롭습니다.
이는 지역 특성, 데이터 산출 방식 등에 따라 달라질 수 있으므로, 단순한 상식과는 다른 결과일 수 있음을 유의해야 합니다.
비용 효율성과 집값의 양(陽)의 상관관계

집값이 높은 지역이 오히려 ‘비용 대비 효용’에서 점수를 잘 받을 가능성이 있음.
즉, 단순한 ‘값싼 지역 = 비용 효율성 높음’이 아니라, 주택 가격을 지불하는 만큼 충분한 가치(인프라, 환경 등)가 있는가가 중요하다는 뜻입니다.

결론적으로
**“안전”**이 종합 평가에서 가장 중요한 키 포인트로 작용하며,
“집값” 자체는 종합 점수를 결정하는 데 큰 변수는 아니지만,
집값이 높은 곳 중에서도 안전성과 편의성, 비용 효율성을 고루 갖춘 지역이 종합 평가가 높을 수 있고,
반대로 집값이 낮아도 안전·편의·비용 효율성을 잘 갖추면 높은 종합 점수를 얻을 수 있습니다.
따라서 지역을 평가·개발하거나 주거 정책을 세울 때, 안전·편의·비용 효율성 등 다각적인 요소를 종합적으로 고려하고, 각 지역의 특성에 맞는 맞춤형 전략이 필요합니다.
